```python
import streamlit as st
import pandas as pd
from sklearn.ensemble import RandomForestClassifier

# ----------------------------
# Sample Training Data
# ----------------------------
data = {
    "Income": [25000, 40000, 50000, 60000, 80000, 100000, 120000, 150000],
    "LoanAmount": [100000, 150000, 200000, 250000, 300000, 350000, 400000, 500000],
    "CreditScore": [550, 620, 680, 720, 760, 800, 850, 900],
    "Approved": [0, 0, 1, 1, 1, 1, 1, 1]
}

df = pd.DataFrame(data)

# Features and Target
X = df[["Income", "LoanAmount", "CreditScore"]]
y = df["Approved"]

# Train Model
model = RandomForestClassifier(
    n_estimators=100,
    random_state=42
)
model.fit(X, y)

# ----------------------------
# Streamlit UI
# ----------------------------
st.set_page_config(
    page_title="Loan Approval Predictor",
    page_icon="💰",
    layout="centered"
)

st.title("💰 Loan Approval Predictor")
st.write("Enter the applicant details below.")

income = st.number_input(
    "Annual Income (₹)",
    min_value=1000,
    max_value=10000000,
    value=50000
)

loan_amount = st.number_input(
    "Loan Amount Required (₹)",
    min_value=1000,
    max_value=10000000,
    value=200000
)

credit_score = st.slider(
    "Credit Score",
    min_value=300,
    max_value=900,
    value=700
)

if st.button("Predict"):
    input_data = pd.DataFrame({
        "Income": [income],
        "LoanAmount": [loan_amount],
        "CreditScore": [credit_score]
    })

    prediction = model.predict(input_data)[0]
    probability = model.predict_proba(input_data)[0]

    st.subheader("Prediction Result")

    if prediction == 1:
        st.success("✅ Loan Likely Approved")
        st.write(f"Approval Confidence: {probability[1]*100:.2f}%")
    else:
        st.error("❌ Loan Likely Rejected")
        st.write(f"Rejection Confidence: {probability[0]*100:.2f}%")

st.markdown("---")
st.caption("Demo project built with Python, Scikit-Learn, and Streamlit.")
```
